```python
from sklearn import linear_model
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\achyu\git\statProjects\SalaryRegression\archive\Salary_dataset.csv")
df
```

```python
%matplotlib inline
plt.xlabel('xp')
plt.ylabel('money')
plt.scatter(df.YearsExperience, df.Salary)
```

```python
regression = linear_model.LinearRegression()
regression.fit(df[['YearsExperience']].values, df.Salary.values)
```

```python
regression.predict([[6.5]])
```

```python
rate = regression.coef_
intercept = regression.intercept_
plt.plot(df.YearsExperience, regression.predict(df[['YearsExperience']]))
plt.xlabel('xp')
plt.ylabel('money')
plt.scatter(df.YearsExperience, df.Salary, color = "red")
```

```python
from sklearn.metrics import mean_squared_error

true = list(df.Salary.values)
xpList = list(df.YearsExperience.values)
predictions = [regression.predict([[i]]) for i in xpList]

rmse = mean_squared_error(true, predictions, squared = False)
print(rmse)
```
