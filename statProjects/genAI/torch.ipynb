```python
import nltk
import spacy
from torch.utils.data import DataLoader, Dataset

# Download required NLP resources
nltk.download("punkt")
spacy.cli.download("en_core_web_sm")

# Tokenization examples
nlp = spacy.load("en_core_web_sm")

def tokenize_text(text):
    return [token.text for token in nlp(text)]

sample_text = "PyTorch is great for NLP tasks!"
print("Tokenized:", tokenize_text(sample_text))
```

```python
import torch
from torch.nn.utils.rnn import pad_sequence

# Example tokenized data
tokenized_texts = [["I", "love", "PyTorch"], ["This", "is", "a", "test"]]

# Convert tokens to numerical values (mock example)
word_to_idx = {word: idx for idx, word in enumerate(set(sum(tokenized_texts, [])))}
numericalized = [[word_to_idx[word] for word in sentence] for sentence in tokenized_texts]

# Pad sequences
padded_sequences = pad_sequence([torch.tensor(seq) for seq in numericalized], batch_first=True)
print("Padded Tensor:\n", padded_sequences)
```
