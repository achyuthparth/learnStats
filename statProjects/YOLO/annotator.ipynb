```python
# Cell 1: Import necessary libraries
import torch
from ultralytics import YOLO


# Load your pretrained model
model_path = r"D:\Users\achyu\git\learnStats\statProjects\YOLO\yolo11n.pt"
model = YOLO(model_path)
```

```python
# Cell 3: Freeze backbone layers of YOLO model (Ultralytics format)
for name, param in model.model.named_parameters():
    if "backbone" in name:
        param.requires_grad = False
```

```python
# Debug: print parameter names
for name, _ in model.model.named_parameters():
    print(name)
```

```python
# Cell 5: Start fine-tuning the model using your dataset
# Path to your dataset YAML file
yaml_path = r"D:\Users\achyu\git\learnStats\statProjects\YOLO\construction.yaml"

# Start fine-tuning
model.train(data=yaml_path, epochs=50, imgsz=640, batch=16)
```

```python
# Evaluate performance on validation set
metrics = model.val()
print(metrics)
```

```python
# Save the best model manually (if needed)
model_path = model.save("C:/yolo_project/transfer_learning/yolo11n_finetuned_1.pt")
print(f"✅ Model saved at: {model_path}")
```

```python
import os
import random
import cv2
import matplotlib.pyplot as plt
from ultralytics import YOLO

# Load your custom fine-tuned YOLO model
model_path = r"E:\yolo_project\transfer_learning\yolo11n_finetuned_1.pt"
model = YOLO(model_path)

# Pick a random image from your dataset
image_dir = r"E:\yolo_project\transfer_learning\initial model\images\train"
img_files = [os.path.join(image_dir, f) for f in os.listdir(image_dir) if f.lower().endswith(('.jpg', '.png'))]
img_path = random.choice(img_files)

# Run inference
results = model(img_path, conf=0.3)

# Plot and display predictions
img = results[0].plot()  # This gives a NumPy array with boxes drawn
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.axis("off")
plt.title(f"Predictions for {os.path.basename(img_path)}")
plt.show()
```
