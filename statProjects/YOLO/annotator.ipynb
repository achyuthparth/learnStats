```python
# Cell 1: Import necessary libraries
import torch
from ultralytics import YOLO


# Load your pretrained model
model_path = r"D:\Users\achyu\git\learnStats\statProjects\YOLO\yolo11n.pt"
model = YOLO(model_path)
```

```python
# Cell 3: Freeze backbone layers of YOLO model (Ultralytics format)
for name, param in model.model.named_parameters():
    if "backbone" in name:
        param.requires_grad = False
```

```python
# Debug: print parameter names
for name, _ in model.model.named_parameters():
    print(name)
```

```python
# Cell 5: Start fine-tuning the model using your dataset
# Path to your dataset YAML file
yaml_path = r"D:\Users\achyu\git\learnStats\statProjects\YOLO\construction.yaml"

# Start fine-tuning
model.train(data=yaml_path, epochs=50, imgsz=640, batch=16)
```

```python
# Evaluate performance on validation set
metrics = model.val()
print(metrics)
```

```python
# Save the best model manually (if needed)
model_path = model.save("C:/yolo_project/transfer_learning/yolo11n_finetuned_1.pt")
print(f"✅ Model saved at: {model_path}")
```

```python
import os
import random
import cv2
import matplotlib.pyplot as plt
from ultralytics import YOLO

# Load your custom fine-tuned YOLO model
model_path = r"E:\yolo_project\transfer_learning\yolo11n_finetuned_1.pt"
model = YOLO(model_path)

# Pick a random image from your dataset
image_dir = r"E:\yolo_project\transfer_learning\initial model\images\train"
img_files = [os.path.join(image_dir, f) for f in os.listdir(image_dir) if f.lower().endswith(('.jpg', '.png'))]
img_path = random.choice(img_files)

# Run inference
results = model(img_path, conf=0.3)

# Plot and display predictions
img = results[0].plot()  # This gives a NumPy array with boxes drawn
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.axis("off")
plt.title(f"Predictions for {os.path.basename(img_path)}")
plt.show()
```

```python
import openai
import os
import keyring

# Set your OpenAI API key
openai.api_key = os.getenv("OPENAI_API_KEY")

messages = [
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "Hello, who won the world series in 2020?"}
]

try:
    response = openai.ChatCompletion.create(
        model="gpt-3.5-turbo",
        messages=messages,
        temperature=0.4,
        max_tokens=1500
    )
    print(response.choices[0].message['content'])
except Exception as e:
    # Retrieve OpenAI API key from keyring if not set in environment
    api_key = os.getenv("OPENAI_API_KEY")
    if not api_key:
        api_key = keyring.get_password("openai_project", "api_key")
        if api_key:
            openai.api_key = api_key
        else:
            print("OpenAI API key not found in environment or keyring.")
    else:
        openai.api_key = api_key
```

```python
detections = []
for box in results[0].boxes:
    cls_id = int(box.cls.item())
    cls_name = model.names[cls_id]
    conf = float(box.conf.item())
    detections.append(f"{cls_name} (confidence: {conf:.2f})")

prompt = (
    "These are objects detected in a construction site image: "
    + ", ".join(detections) + ". "
    + "Based on this, describe what's happening in the scene. "
      "Include potential safety concerns, work activity insights, and environment cues."
)


client = openai.OpenAI(api_key=api_key)
response = client.chat.completions.create(
    model="gpt-4",
    messages=[{"role": "user", "content": prompt}]
)

concise_prompt = (
    "Detected objects in a construction site image: "
    + ", ".join(detections) + ". "
    + "Briefly and factually describe the likely scene, possible safety concerns, and work activity. Be concise and avoid speculation."
)

response = client.chat.completions.create(
    model="gpt-4",
    messages=[{"role": "user", "content": concise_prompt}],
    temperature=0.1
)

print(response.choices[0].message.content)
```
