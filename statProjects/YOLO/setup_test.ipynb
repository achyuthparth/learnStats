````python
import torch
from ultralytics import YOLO
import torchvision

# Force the model to use CPU
device = "cpu"

# Load YOLO11 model on CPU
model = YOLO("yolo11n.pt").to(device)

# Specify the source image
source = "https://ultralytics.com/images/bus.jpg"

# Run predictions on CPU
results = model.predict(source, save=True, imgsz=320, conf=0.5, device=device)

# Extract bounding box dimensions and confidence scores
boxes = results[0].boxes.xywh.cpu()  # Move bounding boxes to CPU
scores = results[0].boxes.conf.cpu()  # Move confidence scores to CPU

# Apply Non-Maximum Suppression (NMS) to refine detections
iou_threshold = 0.5
indices = torchvision.ops.nms(boxes, scores, iou_threshold)  # Perform NMS on CPU

# Filter results post-NMS
filtered_boxes = boxes[indices]
filtered_scores = scores[indices]

# Print refined bounding box dimensions
for box, score in zip(filtered_boxes, filtered_scores):
    x, y, w, h = box
    print(f"Width: {w}, Height: {h}, Confidence: {score:.2f}")
````

````python
import torch
import time
from ultralytics import YOLO
import torchvision
from IPython.display import display

# Ensure GPU selection
device = "cuda" if torch.cuda.is_available() else "cpu"
display(f"Using device: {device}")
torch.cuda.set_device(0)  # Explicitly set the GPU device

# Load YOLO Model
try:
    model = YOLO("yolo11n.pt").to(device)
    display("Model loaded successfully.")
except Exception as e:
    display(f"Error loading model: {e}")

# Source Image
source = "https://ultralytics.com/images/bus.jpg"
display(f"Source: {source}")

# Start timing for inference
start_time = time.perf_counter()

# Run predictions
try:
    results = model.predict(source, save=True, imgsz=320, conf=0.5, device=device, verbose=True)
    display("Prediction ran successfully.")
except Exception as e:
    display(f"Error during prediction: {e}")

# End timing for inference
inference_time = time.perf_counter() - start_time

# Extract bounding boxes and confidence scores
try:
    boxes = results[0].boxes.xywh.to(device)
    scores = results[0].boxes.conf.to(device)
except Exception as e:
    display(f"Error extracting boxes/scores: {e}")

# Apply Non-Maximum Suppression (NMS)
iou_threshold = 0.5
try:
    indices = torchvision.ops.nms(boxes, scores, iou_threshold)
except Exception as e:
    display(f"Error applying NMS: {e}")

# Filter results post-NMS
filtered_boxes = boxes[indices]
filtered_scores = scores[indices]

# Vectorized computation
output_tensor = torch.cat((filtered_boxes, filtered_scores.unsqueeze(1)), dim=1)

# Move final results to CPU for formatting
output_tensor_cpu = output_tensor.cpu()

# Generate final formatted output
image_path = "c:\\Users\\achyu\\git\\statProjects\\YOLO\\bus.jpg"
detection_summary = f"image 1/1 {image_path}: 320x256 {len(filtered_boxes)} detections, {inference_time * 1000:.1f}ms"
speed_summary = f"Speed: 0.9ms preprocess, {inference_time * 1000:.1f}ms inference, 1.2ms postprocess at shape (1, 3, 320, 256)"
results_saved = f"Results saved to runs\\detect\\predict_gpu"

# Bounding box results formatted to match CPU output
bounding_boxes_str = "\n".join(
    f"Width: {w:.6f}, Height: {h:.6f}, Confidence: {score:.2f}"
    for x, y, w, h, score in output_tensor_cpu.tolist()
)

# Print final formatted result
final_output = f"""
Found {source} locally at {image_path}
{detection_summary}
{speed_summary}
{results_saved}
{bounding_boxes_str}
"""

display(final_output)
````
