## Computer Exercise C1

**Problem Statement:**

Use the data in WAGE1.RAW for this exercise.

(i) Find the average education level in the sample. What are the lowest and highest years of education?

(ii) Find the average hourly wage in the sample. Does it seem high or low?

(iii) The wage data are reported in 1976 dollars. Using the Economic Report of the President (2011 or later), obtain and report the Consumer Price Index (CPI) for the years 1976 and 2010.

(iv) Use the CPI values from part (iii) to find the average hourly wage in 2010 dollars. Now does the average hourly wage seem reasonable?

(v) How many women are in the sample? How many men?

**Dataset:** wage1

### Data Loading and Initial Exploration

```python
# Load the WAGE1 dataset (will auto-download if not found)
wage1 = load_wooldridge_data('wage1')

# Display first few rows
print("First 5 rows of WAGE1 dataset:")
print(wage1.head())

print(f"\nDataset shape: {wage1.shape}")
print(f"Number of observations: {len(wage1)}")

# Get column information
print(f"\nColumns in dataset: {list(wage1.columns)}")
```

### Part (i): Education Statistics

```python
# Part (i): Find average, minimum, and maximum education levels
avg_education = wage1['educ'].mean()
min_education = wage1['educ'].min()
max_education = wage1['educ'].max()

print("Education Statistics:")
print(f"Average education level: {avg_education:.2f} years")
print(f"Minimum education level: {min_education} years")
print(f"Maximum education level: {max_education} years")

# Additional statistics for context
print(f"\nAdditional education statistics:")
print(f"Standard deviation: {wage1['educ'].std():.2f} years")
print(f"Median education: {wage1['educ'].median():.1f} years")
```

### Part (ii): Average Hourly Wage

```python
# Part (ii): Find average hourly wage (in 1976 dollars)
avg_wage_1976 = wage1['wage'].mean()

print("Wage Statistics (1976 dollars):")
print(f"Average hourly wage: ${avg_wage_1976:.2f}")
print(f"Minimum wage: ${wage1['wage'].min():.2f}")
print(f"Maximum wage: ${wage1['wage'].max():.2f}")
print(f"Standard deviation: ${wage1['wage'].std():.2f}")

# Assessment: Does it seem high or low?
print(f"\nAssessment:")
print(f"The average wage of ${avg_wage_1976:.2f} per hour in 1976 dollars seems relatively")
print(f"reasonable for that time period, but we need to adjust for inflation to make")
print(f"a proper assessment against today's standards.")
```

### Part (iii): Consumer Price Index Data

```python
# Part (iii): Consumer Price Index values
# Source: Economic Report of the President
# Base year: 1982-84 = 100

cpi_1976 = 56.9  # CPI for 1976
cpi_2010 = 218.1  # CPI for 2010

print("Consumer Price Index (CPI) Values:")
print(f"1976 CPI: {cpi_1976}")
print(f"2010 CPI: {cpi_2010}")
print(f"Base year: 1982-84 = 100")

# Calculate inflation factor
inflation_factor = cpi_2010 / cpi_1976
print(f"\nInflation factor from 1976 to 2010: {inflation_factor:.2f}")
print(f"This means prices increased by {(inflation_factor - 1) * 100:.1f}% from 1976 to 2010")
```

### Part (iv): Inflation-Adjusted Wages

```python
# Part (iv): Convert 1976 wages to 2010 dollars
avg_wage_2010 = avg_wage_1976 * inflation_factor

print("Wage Comparison:")
print(f"Average wage in 1976 dollars: ${avg_wage_1976:.2f}")
print(f"Average wage in 2010 dollars: ${avg_wage_2010:.2f}")

print(f"\nAssessment:")
print(f"The average wage of ${avg_wage_2010:.2f} per hour in 2010 dollars")
print(f"seems reasonable, as it's above the federal minimum wage of $7.25")
print(f"that was in effect in 2010. This represents purchasing power")
print(f"equivalent to what ${avg_wage_1976:.2f} could buy in 1976.")
```

### Part (v): Gender Distribution

```python
# Part (v): Count women and men in the sample
# The 'female' variable is an indicator (1 = female, 0 = male)
num_women = wage1['female'].sum()
num_men = len(wage1) - num_women
total_observations = len(wage1)

print("Gender Distribution:")
print(f"Number of women: {num_women}")
print(f"Number of men: {num_men}")
print(f"Total observations: {total_observations}")

# Calculate percentages
pct_women = (num_women / total_observations) * 100
pct_men = (num_men / total_observations) * 100

print(f"\nPercentages:")
print(f"Women: {pct_women:.1f}%")
print(f"Men: {pct_men:.1f}%")
```

**(vii) Suppose School A spends $6,000 per student and School B spends $5,500 per student. By what percentage does School A's spending exceed School B's? Compare this to 100 · [log(6,000) – log(5,500)], which is the approximation percentage difference based on the difference in the natural logs.**

```python
# Part (vii): Percentage difference calculations
import numpy as np

school_a_spending = 6000
school_b_spending = 5500

# Exact percentage difference
exact_pct_diff = ((school_a_spending - school_b_spending) / school_b_spending) * 100

# Log approximation
log_approx = 100 * (np.log(school_a_spending) - np.log(school_b_spending))

print(f"School A spending: ${school_a_spending}")
print(f"School B spending: ${school_b_spending}")
print(f"Exact percentage difference: {exact_pct_diff:.2f}%")
print(f"Log approximation: {log_approx:.2f}%")
print(f"Difference between methods: {abs(exact_pct_diff - log_approx):.2f} percentage points")
```

## Notes and Key Takeaways

### Important Concepts from Chapter 1:

1. **Descriptive Statistics**: Mean, median, standard deviation, min/max provide insights into data distribution
2. **Inflation Adjustment**: Always adjust monetary values for inflation when comparing across time periods
3. **Missing Data**: Be aware of missing values and how they affect sample sizes in calculations
4. **Misleading Averages**: When data has many zeros, the overall average may not represent the "typical" non-zero case
5. **Data Exploration**: Always examine your data before running regressions

### Key Formulas:

- **Inflation Adjustment**: Real Value = Nominal Value × (CPI_target / CPI_original)
- **Percentage Change**: % Change = [(New - Old) / Old] × 100
- **Log Approximation**: % Change ≈ 100 × [log(New) - log(Old)]
